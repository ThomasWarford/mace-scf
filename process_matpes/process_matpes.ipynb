# Process MatPES

Download the MatPES 2025.2 datasets (r2SCAN and PBE) from Hugging Face and load them for processing.

```python
import urllib.request
from pathlib import Path

from monty.serialization import loadfn

DATA_DIR = Path("data")
DATA_DIR.mkdir(exist_ok=True)

URLS = {
    "r2scan": "https://huggingface.co/datasets/materialyze/matpes/resolve/main/MatPES-R2SCAN-2025.2.jsonl?download=true",
    "pbe": "https://huggingface.co/datasets/materialyze/matpes/resolve/main/MatPES-PBE-2025.2.jsonl?download=true",
}
```

```python
def download(url: str, dest: Path) -> Path:
    if dest.exists():
        print(f"{dest} already exists, skipping download")
        return dest
    print(f"Downloading {url} -> {dest}")
    urllib.request.urlretrieve(url, dest)
    print(f"Done ({dest.stat().st_size / 1e6:.1f} MB)")
    return dest


paths = {
    "r2scan": download(URLS["r2scan"], DATA_DIR / "MatPES-R2SCAN-2025.2.jsonl"),
    "pbe": download(URLS["pbe"], DATA_DIR / "MatPES-PBE-2025.2.jsonl"),
}
paths
```
